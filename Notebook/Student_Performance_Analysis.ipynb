```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv(r"C:\Users\RAKSHITHA K C\Downloads\StudentsPerformance\StudentsPerformance.csv")
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df["Average Marks"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
) / 3
```

```python
df.head()
```

```python
df["Average Marks"].max()
```

```python
df["Average Marks"].min()
```

```python
df["Average Marks"].mean()
```

```python
plt.figure(figsize=(6,4))

df.groupby("gender")["Average Marks"].mean().plot(kind="bar")

plt.title("Average Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")

plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.hist(df["Average Marks"], bins=10)

plt.title("Average Marks Distribution")
plt.xlabel("Average Marks")
plt.ylabel("Students")

plt.show()
```

```python
df["gender"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Gender Distribution")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.show()
```

## Insights

- Most students scored between 60 and 80.
- Female students performed slightly better than male students.
- Reading and writing scores are strongly correlated.
- Only a few students scored below 40.
